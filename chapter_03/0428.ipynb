4. 문자열 s를 입력받아, 그 문자열이 회문인지 여부를 확인하는 함수를 작성하세요.
회문(Palindrome)은 앞뒤가 똑같이 읽히는 문자열을 의미합니다. 예를 들어, "racecar"나 "madam"은 회문이지만, "hello"나 "world"는 아닙니다.

```python
def is_palindrome(s: str) -> bool:
   return s == s[::-1]

input= ["racecar", "madam", "hello", "level", "world", "noon", "python"]
output = [True, True, False, True, False, True, False]
for input_value, output_value in zip(input, output):
   assert is_palindrome(input_value) == output_value
   print(f"입력값: {input_value} ➔ 판별결과: {is_palindrome(input_value) == output_value}")
```

6. 문자열 s를 입력받아, 해당 문자열을 뒤집고, 그 문자열에 포함된 모음(a, e, i, o, u)의 개수를 반환하는 두 가지 함수를 작성하세요.

```python
def reverse(s: str) -> str:
  return s[::-1]

def count_vowels(s :str) -> int:
  vowel = ['a', 'e', 'i', 'o', 'u']
  count = 0
  for char in s:
    if char.lower() in vowel:
      count += 1
  return count
  
  
input_data_string_reversal_vowel_count = ["hello", "world", "python", "aeiou", "empty", "hElLo"]
output_data_string_reversal_vowel_count = [("olleh", 2), ("dlrow", 1), ("nohtyp", 1), ("uoiea", 5), ("ytpme", 1), ("oLlEh", 2)]

for input_value, output_value in zip(input_data_string_reversal_vowel_count, output_data_string_reversal_vowel_count):
    assert (reverse(input_value), count_vowels(input_value)) == output_value
    print(f"입력값: {reverse(input_value), count_vowels(input_value)} ➔ 판별결과: {(reverse(input_value), count_vowels(input_value)) == output_value}")
```

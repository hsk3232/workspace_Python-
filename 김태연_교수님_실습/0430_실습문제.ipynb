## 실습 문제 1: ReLU 함수 (Rectified Linear Unit)
ReLU 함수 ( f(x) = \max(0, x) )는 딥러닝에서 가장 널리 사용되는 비선형 활성화 함수입니다. 입력값이 0보다 작으면 0을, 0 이상이면 입력값 그대로를 출력합니다. 이 함수의 동작을 시각화해보세요.

{\displaystyle \operatorname {ReLU} (x)=x^{+}=\max(0,x)={\frac {x+|x|}{2}}={\begin{cases}x&{\text{if }}x>0,\\0&x\leq 0\end{cases}}}

요구사항:
( x )의 범위를 -5부터 5까지 설정하고, NumPy를 사용해 ReLU 함수 값을 계산하세요.
Matplotlib으로 그래프를 그리고, x축 레이블을 "x", y축 레이블을 "f(x)", 제목을 "ReLU Function"으로 설정하세요.
그래프에서 ( x = 0 )에서의 꺾이는 지점을 관찰하세요.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. x 범위 설정: -5부터 5까지
x = np.linspace(-5, 5, 100)

# 2. ReLU 함수 정의 및 적용
def relu(x):
    return np.maximum(0, x)

y = relu(x)

# 3. 그래프 그리기
plt.plot(x, y, label='ReLU(x)', color='blue')
plt.title("ReLU Function")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--')  # x = 0의 꺾이는 지점 강조
plt.legend()
plt.show()
```

## 실습 문제 2: 평균 제곱 오차 손실 함수 (Mean Squared Error Loss)
{\displaystyle \operatorname {MSE} ={\frac {1}{n}}\sum _{i=1}^{n}\left(Y_{i}-{\hat {Y_{i}}}\right)^{2}}

return  sum ((a-b)**2) /n

MSE 손실 함수는 예측값과 실제값 간의 오차를 측정합니다. 실제값을 고정하고 예측값을 변화시키며 손실의 변화를 시각화해보세요.

요구사항:
실제값 ( y = 1 )로 고정하고, 예측값 ( \hat{y} )를 -4부터 4까지 변화시키며 MSE를 계산하세요.
Matplotlib으로 그래프를 그리고, x축 레이블을 "Predicted Value", y축 레이블을 "Loss", 제목을 "Mean Squared Error Loss"로 설정하세요.
최소 손실 지점(( \hat{y} = 1 ))을 빨간 점으로 표시하세요.

```python

# 1. x 범위 설정
x = np.linspace(-4, 4, 100)

# 2. y 값 고정
y = 1

# 3. MSE 함수 사용
def MSE(x, y):
    return ((x-y)**2)

# 4. 그래프 그리기
plt.plot(x, MSE(x, y), color='blue')
plt.title("Mean Squared Error Loss")
plt.xlabel("Predicted Value")
plt.ylabel("Loss")
plt.grid(True)
# 5. 최소 손실 지점 표시 (예측값 = 실제값 = 1)
plt.scatter([1], [0], color='red', label='Minimum Loss (ŷ=1)')

plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 예측값: -4부터 4까지 100개 생성
y_pred = np.linspace(-4, 4, 100)

# 2. 실제값: 1을 100개 생성
y_true = np.ones(100)  # 또는 y_true = np.array([1]*100)

# 3. MSE 함수 (교수님 방식)
def mse(y_true, y_pred):
    n = len(y_true)
    return np.sum((y_true - y_pred) ** 2) / n

# 4. 각 예측값에 대해 MSE 계산
loss = (y_pred - y_true) ** 2  # 길이 100인 배열 반환


# 5. 시각화
plt.plot(y_pred, loss, label='MSE Loss', color='blue')
plt.scatter([1], [0], color='red', label='Minimum Loss (ŷ = 1)')
plt.title("Mean Squared Error Loss")
plt.xlabel("Predicted Value")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()
plt.show()
```

add_item(item_name: str, price: float, quantity: int): 이 메서드는 지정된 수량만큼 아이템을 카트에 추가합니다. 만약 이미 카트에 있는 아이템이라면, 해당 아이템의 수량을 증가시킵니다.

remove_item(item_name: str, quantity: int): 카트에서 지정된 수량만큼 아이템을 제거합니다. 제거할 수량이 현재 남아있는 수량보다 크거나 같으면, 해당 아이템을 카트에서 완전히 삭제합니다.

get_total_price(): 카트에 있는 모든 아이템을 순회하며 각 아이템의 가격과 수량을 곱해 총 비용을 계산합니다.

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item_name: str, price: float, quantity: int):
        pass

    def remove_item(self, item_name: str, quantity: int):
        pass

    def get_total_price(self) -> float:
        pass


input_data_cart = [
    [("add", "apple", 1.5, 10), ("add", "banana", 0.5, 5), ("add", "apple", 1.5, 5)],  # Adding items, updating apple
    [("remove", "banana", None, 2), ("remove", "apple", None, 12), ("add", "orange", 2.0, 2)],  # Removing and adding more
]

expected_cart_states = [
    {"apple": (1.5, 15), "banana": (0.5, 5)},
    {"orange": (2.0, 2)}
]

expected_total_prices = [25.0, 4.0]

for data_set, expected_cart, expected_total in zip(input_data_cart, expected_cart_states, expected_total_prices):
    cart = ShoppingCart()

    for action, item_name, price, quantity in data_set:
        if action == "add":
            cart.add_item(item_name, price, quantity)
        elif action == "remove":
            cart.remove_item(item_name, quantity)

    assert cart.items == expected_cart

    total_price = cart.get_total_price()
    assert total_price == expected_total
```

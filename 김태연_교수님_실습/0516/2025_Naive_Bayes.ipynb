### 실습 문제 1: 동전 던지기 확률 계산하기

$$ P(A) = \frac{\text{사건 A에 유리한 결과의 수}}{\text{표본 공간의 총 결과의 수}} $$

Python을 사용하여 앞면(Head)과 뒷면(Tail)이 나올 확률이 같은 공정한 동전을 100번 던졌을 때, 앞면과 뒷면이 각각 몇 번 나오는지 시뮬레이션하고, 각 면이 나올 실험적 확률을 계산하세요. Matplotlib 를 사용하여 이 확률 분포를 히스토그램으로 시각화하세요.



```python
import random
import matplotlib.pyplot as plt

plt.rc('font', family='D2Coding')

# 1. 동전 던지기 시뮬레이션
trials = 100
results = [random.choice(['Head', 'Tail']) for _ in range(trials)]

# 2. 결과 세기
head_count = results.count('Head')
tail_count = results.count('Tail')

# 3. 실험적 확률 계산
head_prob = head_count / trials
tail_prob = tail_count / trials

# 4. 결과 출력
print(f"🪙 앞면: {head_count}회, 확률: {head_prob:.2f}")
print(f"🪙 뒷면: {tail_count}회, 확률: {tail_prob:.2f}")

# 5. 히스토그램 그리기
plt.bar(['Head', 'Tail'], [head_prob, tail_prob], color=['skyblue', 'lightcoral'])
plt.title("동전 던지기 실험 확률 분포")
plt.ylabel("확률")
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### 실습 문제 2: 주사위 던지기 독립 사건(Independent Events) 확률 계산하기

두 사건 A와 B가 독립일 때, A와 B가 동시에 발생할 확률은 다음과 같이 계산됩니다:

$$ P(A \cap B) = P(A) \cdot P(B) $$

예를 들어, 두 개의 주사위를 던져 첫 번째 주사위가 6이 나올 확률 $P(6) = \frac{1}{6} $, 두 번째 주사위를 던져 6이 나올 확률 $ P(6) = \frac{1}{6} $ 두 사건이 독립이므로 동시에 발생할 확률은:

$$ P(6 \cap 6) = \frac{1}{6} \cdot \frac{1}{6} = \frac{1}{36} $$

Python을 사용하여 두 개의 주사위를 1000번 던졌을 때, 각 눈금(2부터 12)이 나오는 횟수를 시뮬레이션하고, 각 눈금이 나올 실험적 확률을 계산하세요. Matplotlib 를 사용하여 이 확률 분포를 히스토그램으로 시각화하세요.


```python
import random
import matplotlib.pyplot as plt

plt.rc('font', family='D2Coding')

# 1. 시뮬레이션: 두 개의 주사위 던지기 1000번
trials = 1000
sums = []

for _ in range(trials):
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)
    sums.append(die1 + die2)

# 2. 결과 집계
sum_counts = {i: 0 for i in range(2, 13)}
for s in sums:
    sum_counts[s] += 1

# 3. 확률 계산
sum_probs = {k: v / trials for k, v in sum_counts.items()}

# 4. 시각화
plt.bar(sum_probs.keys(), sum_probs.values(), color='lightgreen')
plt.title("🎲 두 개의 주사위 합의 실험적 확률 분포")
plt.xlabel("합 (2 ~ 12)")
plt.ylabel("확률")
plt.xticks(range(2, 13))
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### 실습 문제 3: 조건부 확률 계산하기

조건부 확률은 어떤 사건이 일어났다는 조건 하에 다른 사건이 일어날 확률입니다.

$$ P(A|B) = \frac{P(A \cap B)}{P(B)} $$

다음과 같은 학생 설문조사 데이터가 있습니다:

| 학생 ID | 전공 | 학년 | 과외활동 참여 |
| --- | --- | --- | --- |
| 1 | 공학 | 1 | Y |
| 2 | 인문 | 2 | N |
| 3 | 공학 | 3 | Y |
| 4 | 상경 | 2 | N |
| 5 | 인문 | 1 | Y |
| 6 | 공학 | 4 | N |
| 7 | 상경 | 3 | Y |
| 8 | 공학 | 2 | Y |

위 데이터를 Pandas DataFrame으로 만들고, 다음 확률들을 계산하세요:

- P(전공=공학): 전체 학생 중 공학 전공 학생의 확률
- P(과외활동=예): 전체 학생 중 과외활동에 참여하는 학생의 확률
- P(전공=공학∣과외활동=예): 과외활동에 참여하는 학생 중 공학 전공 학생의 확률
- P(과외활동=예∣전공=공학): 공학 전공 학생 중 과외활동에 참여하는 학생의 확률

```python
import pandas as pd

# 학생 설문조사 데이터
data = {
    '학생ID': [1, 2, 3, 4, 5, 6, 7, 8],
    '전공': ['공학', '인문', '공학', '상경', '인문', '공학', '상경', '공학'],
    '학년': [1, 2, 3, 2, 1, 4, 3, 2],
    '과외활동': ['Y', 'N', 'Y', 'N', 'Y', 'N', 'Y', 'Y']
}

df = pd.DataFrame(data)
```

```python
# 전체 학생 수
total = len(df)
```

```python
# P(전공=공학)
p_engineering = len(df[df['전공'] == '공학']) / total
```

```python
# P(과외활동=예)
p_activity_yes = len(df[df['과외활동'] == 'Y']) / total
```

```python
# P(전공=공학 ∣ 과외활동=예)
activity_yes = df[df['과외활동'] == 'Y']
p_engineering_given_activity = len(activity_yes[activity_yes['전공'] == '공학']) / len(activity_yes)
```

```python
# P(과외활동=예 ∣ 전공=공학)
engineering = df[df['전공'] == '공학']
p_activity_given_engineering = len(engineering[engineering['과외활동'] == 'Y']) / len(engineering)
```

```python
print(f"P(전공=공학) = {p_engineering:.2f}")
print(f"P(과외활동=예) = {p_activity_yes:.2f}")
print(f"P(전공=공학 | 과외활동=예) = {p_engineering_given_activity:.2f}")
print(f"P(과외활동=예 | 전공=공학) = {p_activity_given_engineering:.2f}")
```

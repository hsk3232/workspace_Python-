# 과제 8: Iris 데이터셋으로 Naive Bayes 분류 모델 만들기

Iris 데이터셋을 사용해 꽃의 종류를 Naive Bayes 모델로 분류하고, 결과를 시각화합니다.

## 1. Iris 데이터셋 다운로드 및 로드

Scikit-learn에 포함된 Iris 데이터셋을 불러옵니다. 이 데이터셋은 꽃잎과 꽃받침의 길이/너비를 기반으로 3종류의 Iris 꽃을 분류합니다. (sklearn.datasets.load_iris)

## 2. 데이터프레임 만들기

Iris 데이터를 pandas DataFrame으로 변환해 데이터를 보기 쉽게 정리합니다. 특성(feature)과 타겟(target)을 포함합니다. (iris.data, iris.feature_names, iris.target)

## 3. 데이터 확인

데이터의 구조를 확인하고, 첫 5행을 출력해 데이터가 어떻게 생겼는지 확인합니다.

## 4. 데이터 분할 (훈련/테스트)

데이터를 훈련용(80%)과 테스트용(20%)으로 나눠 모델을 학습하고 평가할 준비를 합니다. (sklearn.model_selection.train_test_split, X_train, X_test, y_train, y_test)

## 5. Naive Bayes 모델 생성

Gaussian Naive Bayes 모델을 생성합니다. 이 모델은 features 가 연속형인 데이터에 적합합니다. (sklearn.naive_bayes.GaussianNB)

## 6. 모델 학습

훈련 데이터를 사용해 Naive Bayes 모델을 학습시킵니다. (fit)

## 7. 예측 수행

테스트 데이터를 사용해 모델이 꽃의 종류를 예측합니다. (predict)

## 8. 모델 평가

모델의 정확도를 계산해 성능을 확인합니다. (sklearn.metrics.accuracy_score)

```python
import math
import numpy as np
import matplotlib.pyplot as plt

# 1. 순수 Python 버전
x_py = [i * 0.1 for i in range(int((2 * math.pi) / 0.1) + 1)]
y_py = [math.sin(x) for x in x_py]

# 그래프 그리기
plt.plot(x_py, y_py, label='math.sin')
plt.show()
```

```python

# 2. NumPy 버전
x_np = np.linspace(0, 2 * np.pi, 100)
y_np = np.sin(x_np)

# 그래프 그리기
plt.plot(x_np, y_np, label='np.sin')
plt.show()
```

### 16.1.8 두 번째 데이터 추가하기

```python
from pathlib import Path
import csv
from datetime import datetime as dt
import matplotlib.pyplot as plt

path = Path('the_csv_file_format\partial_programs\weather_data\sitka_weather_2021_simple.csv')
lines = path.read_text().splitlines()

reader = csv.reader(lines)
header_row = next(reader)
    
# 날짜와 최고 기온 추출
dates, highs, lows = [], [], []
for row in reader:
    current_date = dt.strptime(row[2], '%Y-%m-%d')
    high = int(row[4])
    low = int(row[5])
    dates.append(current_date)
    highs.append(high)
    lows.append(low)

# 최고 기온을 그래프로 그림
plt.style.use('seaborn-v0_8')
fig, ax = plt.subplots()
ax.plot(dates, highs, color='red', alpha=0.5)
ax.plot(dates, lows, color='blue', alpha=0.5)
ax.fill_between(dates, highs, lows, facecolor='blue', alpha=0.1)

# 그래프 형식
ax.set_title("Daily High and Low Temperatures, 2021", fontsize=24)
ax.set_xlabel('', fontsize=16)
fig.autofmt_xdate()
ax.set_ylabel("Temperature (F)", fontsize=16)
ax.tick_params(labelsize=16)
plt.show()
```

```python
import csv
from dataclasses import dataclass
import pandas as pd
import plotly.express as px
from pathlib import Path
from datetime import datetime


@dataclass
class Country:
    name: str
    gdp: float
    life_expectancy: float
    density: float
    lat: float
    lon: float


class CountryStatistics:
    # 생성자
    def __init__(self):
        self.countries = []
        self.load_time = None

    # 문자열을 입력으로 받아 쉼표, 달러 기호, 백분율 기호를 제거하고 숫자 값을 float으로 반환
    def __clean_numeric_data(self, value: str) -> float:
        value = value.strip().replace(",", "").replace("$", "").replace("%", "")
        if value == '':
            return 0.0  # 빈칸은 0으로
        return float(value)  # 숫자로 변환 실패 시 오류 발생

    # 데이터 로드 및 파싱, 정리된 데이터는 self.countries 리스트에 저장
    def __load_data(self, filename: str):
        path = Path(filename)
        with path.open(encoding='utf-8') as f:
            reader = csv.DictReader(f)
            
            for row in reader:
                try:
                    name = row['Country']
                    density = self.__clean_numeric_data(row['Density (P/Km2)'])
                    gdp = self.__clean_numeric_data(row['GDP'])
                    life = self.__clean_numeric_data(row['Life expectancy'])
                    lat = self.__clean_numeric_data(row['Latitude'])
                    lon = self.__clean_numeric_data(row['Longitude'])

                    country = Country(name, gdp, life, density, lat, lon)
                    self.countries.append(country)
                except Exception as e:
                    print(f"❌ 오류 발생: {e}, 행: {row}")
                    continue
        
        self.load_time = datetime.now()


    def top_5_gdp(self):
        sorted_list = sorted(self.countries, key=lambda c: c.gdp, reverse=True)[:5]
        return [(c.name, f"{c.gdp:,.0f}") for c in sorted_list]
    
    def top_5_life_expectancy(self): 
        sorted_life = sorted(self.countries, key=lambda c: c.life_expectancy, reverse=True)[:5]
        return [(c.name, f"{c.life_expectancy:,.0f}") for c in sorted_life]
        
    def countries_by_density(self):
        sorted_density = sorted(self.countries, key=lambda c: c.density, reverse=True)[:5]
        return [(c.name, f"{c.density:,.0f}") for c in sorted_density]

    def get_data_load_time(self, format_str=None):
        if self.load_time is None:
            return "아직 데이터를 로드하지 않았습니다."

        if format_str:
            return self.load_time.strftime(format_str)

    def plot_geo_scatter(self, color_by='GDP', size_by='GDP'):
        df = pd.DataFrame([{
            'Country': c.name,
            'GDP': c.gdp,
            'Life Expectancy': c.life_expectancy,
            'Density': c.density,
            'Latitude': c.lat,
            'Longitude': c.lon
        } for c in self.countries])

        fig = px.scatter_geo(
            df,
            lat='Latitude',
            lon='Longitude',
            text='Country',
            color=color_by,
            size=size_by,
            projection='natural earth',
            title=f"🌍 {color_by} 기반 세계 시각화"
        )
        fig.show()

        

          
s = CountryStatistics()
s._CountryStatistics__load_data('world-data-2023.csv')  # 비공개 메서드라서 이렇게 접근
print(s.top_5_gdp())

print(s.top_5_life_expectancy())
print(s.get_data_load_time("%Y-%m-%d %H:%M:%S"))  # 포맷 지정 출력
s.plot_geo_scatter()

```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def get_x(self):
        return self.x
    def get_y(self):
        return self.y
    
    def set_x(self, x):
        self.x = x
    
    def set_y(self, y):
        self.y = y
    
p = Point(1, 2)
p.set_x(5)
p.set_y(10)
p.get_x(), p.get_y()
```

```python
class Rectangle(Point):
    def __init__(self, x, y):
        super().__init__(x, y)
    
r = Rectangle(1, 2)
print(type(r))
```

```python
class Person:
    def __init__(self, name):
        self.name = name
    
    def display(slef):
        msg = f"저는 {slef.name} 입니다."
        print(msg)
        
p = Person("홍지민😆")
p.display()
```

```python
class Student(Person):
    def __init__(self, name, student_id):
        super().__init__(name)
        self.student_id = student_id
    
    def get_student_id(self):
        return self.student_id
    
    def display(self):
        # 코드 재활용을 위해서 아래처럼 쓰는 것은 비효율 적임.
        # msg = f"저는 {self.name} 입니다. \n" + f"학번은{self.student_id} 입니다."
        super().display()
        # msg = f"학번은{self.student_id} 입니다."
        msg = f"학번은{self.get_student_id()} 입니다." # 클래스 안에서 함수 부르기 가능 
        print(msg)
    


s = Student("홍지민😆", '202555100')
s.display()
```

```python
from abc import ABC

class Person(ABC):
    def __init__(self, name):
        self.name = name

    @abc.abstractmethod
    def display(self):
        pass
    
    
```
